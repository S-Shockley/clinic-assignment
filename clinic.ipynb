## Task 4

```python
median_age = clinic["age_years"].median()
print(median_age)
clinic['age_years'] = clinic['age_years'].fillna(median_age)
```

```python
print(clinic["age_years"].isna().sum())
```

## Task 5

```python
clinic["visit_cost_usd"] = clinic["visit_cost_usd"].fillna(
    clinic.groupby("visit_reason")["visit_cost_usd"]
    .transform("median")
)
```

```python
clinic["visit_cost_usd"].isna().sum()
```

There are different costs associated with each visit reason. This is why you would group by that column.
